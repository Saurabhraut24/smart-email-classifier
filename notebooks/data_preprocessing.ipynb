```python
import pandas as pd

df = pd.read_csv('../data/raw/rows.csv/rows.csv')
df.head()
```

```python
df = df[['Product', 'Issue', 'Consumer complaint narrative']]
df.head()
```

```python
df = df.dropna(subset=['Consumer complaint narrative'])
df.shape
```

```python
df = df.rename(columns={
    'Consumer complaint narrative': 'text',
    'Product': 'category',
    'Issue': 'subcategory'
})

df.head()
```

```python
import re

def clean_text(t):
    t = str(t).lower()                              # lowercase
    t = re.sub(r'<.*?>', ' ', t)                    # remove HTML tags
    t = re.sub(r'http\S+|www\S+', ' ', t)           # remove URLs
    t = re.sub(r'[^a-zA-Z ]', ' ', t)               # remove numbers + punctuation
    t = re.sub(r'\s+', ' ', t).strip()              # remove extra spaces
    return t

df['clean_text'] = df['text'].apply(clean_text)
df[['text', 'clean_text']].head()
```

```python
import nltk
from nltk.corpus import stopwords

nltk.download('stopwords')
stop_words = set(stopwords.words('english'))

def remove_stopwords(t):
    return " ".join([word for word in t.split() if word not in stop_words])

df['clean_text'] = df['clean_text'].apply(remove_stopwords)

df[['text', 'clean_text']].head()
```

```python
df.to_csv("cleaned_complaints.csv", index=False)
```
